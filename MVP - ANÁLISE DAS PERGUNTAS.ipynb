# 1 - Como o intercâmbio afeta o preço?

```python
%sql
SELECT
    id_subsistema,
    papel_intercambio,
    ROUND(SUM(pld_medio * qtd_horas) / SUM(qtd_horas), 2)              AS pld_medio,
    ROUND(SUM(spread_medio_se * qtd_horas) / SUM(qtd_horas), 2)        AS spread_medio_se,
    ROUND(SUM(pct_horas_descoladas * qtd_horas) / SUM(qtd_horas), 2)   AS pct_horas_descoladas
FROM portfolio_energia.gold.agg_intercambio_preco
GROUP BY id_subsistema, papel_intercambio
ORDER BY id_subsistema, papel_intercambio;
```

# 2 - Qual fonte mais impacta o PLD?

```python
%sql
SELECT
    fonte,
    ROUND(SUM(desvio_medio_alta  * qtd_horas) / SUM(qtd_horas), 2)  AS desvio_geracao_alta,
    ROUND(SUM(desvio_medio_baixa * qtd_horas) / SUM(qtd_horas), 2)  AS desvio_geracao_baixa,
    ROUND(SUM(ABS(impacto) * qtd_horas) / SUM(qtd_horas), 2)        AS impacto_medio
FROM portfolio_energia.gold.agg_impacto_fonte
GROUP BY fonte
ORDER BY impacto_medio DESC;
```

# 3 - Qual fonte mais impacta em cada submercado?

```python
%sql
SELECT
    id_subsistema,
    fonte,
    ROUND(SUM(ABS(impacto) * qtd_horas) / SUM(qtd_horas), 2) AS impacto_medio
FROM portfolio_energia.gold.agg_impacto_fonte
GROUP BY id_subsistema, fonte
ORDER BY id_subsistema, impacto_medio DESC;
```
